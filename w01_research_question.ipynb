## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np
import os

# Try common paths – change this if yours is different
paths = [
    '../../data/starter_dataset.csv',
    '../data/starter_dataset.csv',
    'data/starter_dataset.csv'
]

df = None
for p in paths:
    if os.path.exists(p):
        df = pd.read_csv(p)
        print(f"✅ Loaded: {p}")
        break

if df is None:
    print("❌ File not found. Please update the path.")
else:
    # Automatically pick a numeric target (skip 'id' columns)
    num_cols = df.select_dtypes(include=[np.number]).columns
    target = [c for c in num_cols if 'id' not in c.lower()][0]

    # -------- 3 NUMBERS THAT MAKE THIS WORTHWHILE --------

    # 1. Row count – enough data for statistical power?
    print(f"\n📊 1. Total records: {len(df):,}")
    print(f"   → More than 5K rows means our correlation tests will be statistically significant.")

    # 2. Missing rate in target – can we trust it?
    missing = df[target].isnull().mean()
    print(f"\n📊 2. Missing rate in '{target}': {missing:.1%}")
    print(f"   → Below 40% means we have enough complete cases to work with.")

    # 3. Coefficient of Variation (CV) – is there enough variance to find signals?
    cv = df[target].std() / (df[target].mean() + 1e-9)
    print(f"\n📊 3. Coefficient of Variation (CV) of '{target}': {cv:.3f}")
    print(f"   → CV > 0.5 means high variance – there's real variation to explain, so ranking signals will be meaningful and actionable.")

    print("\n✅ These numbers confirm: enough data, clean enough target, and real variance → this lane is worth 7 weeks.")
```
